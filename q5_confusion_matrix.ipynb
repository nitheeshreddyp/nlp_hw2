```python
# Q5: Evaluation Metrics from a Multi-Class Confusion Matrix
# This program calculates:
# 1. Precision and recall for each class
# 2. Macro-averaged precision and recall
# 3. Micro-averaged precision and recall


# Confusion matrix
# Rows = System Predictions
# Columns = Gold/Actual Classes
#
#             Gold
#             Cat  Dog  Rabbit
# System Cat    5   10     5
# System Dog   15   20    10
# System Rabbit 0   15    10

confusion_matrix = [
    [5, 10, 5],
    [15, 20, 10],
    [0, 15, 10]
]

classes = ["Cat", "Dog", "Rabbit"]


# Store precision and recall values
precisions = []
recalls = []


# Calculate precision and recall for each class
for i in range(len(classes)):

    # True Positive
    tp = confusion_matrix[i][i]

    # False Positive
    fp = sum(confusion_matrix[row][i] for row in range(len(classes))) - tp

    # False Negative
    fn = sum(confusion_matrix[i]) - tp

    # Precision
    precision = tp / (tp + fp)

    # Recall
    recall = tp / (tp + fn)

    precisions.append(precision)
    recalls.append(recall)

    print(classes[i])
    print("  True Positive:", tp)
    print("  False Positive:", fp)
    print("  False Negative:", fn)
    print("  Precision:", round(precision, 4))
    print("  Recall:", round(recall, 4))
    print()


# Macro-averaged precision
macro_precision = sum(precisions) / len(precisions)

# Macro-averaged recall
macro_recall = sum(recalls) / len(recalls)


# Calculate total True Positives
total_tp = sum(confusion_matrix[i][i] for i in range(len(classes)))

# Calculate total number of examples
total_examples = sum(sum(row) for row in confusion_matrix)


# Micro-averaged precision
micro_precision = total_tp / total_examples

# Micro-averaged recall
micro_recall = total_tp / total_examples


# Print final results
print("===================================")
print("FINAL RESULTS")
print("===================================")

print("Macro Precision:", round(macro_precision, 4))
print("Macro Recall:", round(macro_recall, 4))

print("Micro Precision:", round(micro_precision, 4))
print("Micro Recall:", round(micro_recall, 4))
```
