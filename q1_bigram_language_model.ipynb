```python
# Part II Q1: Bigram Language Model
# This program:
# 1. Reads the training corpus
# 2. Calculates unigram counts
# 3. Calculates bigram counts
# 4. Calculates bigram probabilities using MLE
# 5. Calculates the probability of a sentence
# 6. Compares two test sentences


from collections import Counter


# --------------------------------------------------
# 1. Training Corpus
# --------------------------------------------------

corpus = [
    ["<s>", "I", "love", "NLP", "</s>"],
    ["<s>", "I", "love", "deep", "learning", "</s>"],
    ["<s>", "deep", "learning", "is", "fun", "</s>"]
]


# --------------------------------------------------
# 2. Calculate Unigram Counts
# --------------------------------------------------

unigram_counts = Counter()

for sentence in corpus:
    for word in sentence:
        unigram_counts[word] += 1


# --------------------------------------------------
# 3. Calculate Bigram Counts
# --------------------------------------------------

bigram_counts = Counter()

for sentence in corpus:
    for i in range(len(sentence) - 1):
        bigram = (sentence[i], sentence[i + 1])
        bigram_counts[bigram] += 1


# --------------------------------------------------
# 4. Calculate Bigram Probabilities using MLE
# --------------------------------------------------

bigram_probabilities = {}

for (word1, word2), count in bigram_counts.items():

    # Number of times word1 appears
    denominator = unigram_counts[word1]

    probability = count / denominator

    bigram_probabilities[(word1, word2)] = probability


# --------------------------------------------------
# 5. Print Unigram Counts
# --------------------------------------------------

print("===================================")
print("UNIGRAM COUNTS")
print("===================================")

for word, count in unigram_counts.items():
    print(word, ":", count)


# --------------------------------------------------
# 6. Print Bigram Counts
# --------------------------------------------------

print()
print("===================================")
print("BIGRAM COUNTS")
print("===================================")

for bigram, count in bigram_counts.items():
    print(bigram, ":", count)


# --------------------------------------------------
# 7. Print Bigram Probabilities
# --------------------------------------------------

print()
print("===================================")
print("BIGRAM PROBABILITIES")
print("===================================")

for bigram, probability in bigram_probabilities.items():
    print(bigram, ":", round(probability, 4))


# --------------------------------------------------
# 8. Function to Calculate Sentence Probability
# --------------------------------------------------

def sentence_probability(sentence):

    probability = 1.0

    for i in range(len(sentence) - 1):

        word1 = sentence[i]
        word2 = sentence[i + 1]

        # Get the bigram probability
        bigram_probability = bigram_probabilities.get(
            (word1, word2), 0
        )

        probability *= bigram_probability

    return probability


# --------------------------------------------------
# 9. Test Sentence 1
# --------------------------------------------------

sentence1 = [
    "<s>", "I", "love", "NLP", "</s>"
]

probability1 = sentence_probability(sentence1)


# --------------------------------------------------
# 10. Test Sentence 2
# --------------------------------------------------

sentence2 = [
    "<s>", "I", "love", "deep", "learning", "</s>"
]

probability2 = sentence_probability(sentence2)


# --------------------------------------------------
# 11. Print Sentence Probabilities
# --------------------------------------------------

print()
print("===================================")
print("SENTENCE PROBABILITIES")
print("===================================")

print("S1:", probability1)
print("S2:", probability2)


# --------------------------------------------------
# 12. Compare the Two Sentences
# --------------------------------------------------

print()
print("===================================")
print("MODEL PREFERENCE")
print("===================================")

if probability1 > probability2:

    print("The model prefers S1.")
    print("S1 has the higher probability.")

elif probability2 > probability1:

    print("The model prefers S2.")
    print("S2 has the higher probability.")

else:

    print("Both sentences have the same probability.")
```
